```python
import os
import urllib.request
import zipfile

quarters = [
    "2022-06",
    "2022-09",
    "2022-12",
    "2023-03",
    "2023-06",
    "2023-09",
    "2023-12",
    "2024-03",
    "2024-06",
    "2024-09",
    "2024-12",
    "2025-03",
    "2025-06",
    "2025-09",
    "2025-12",
    "2026-03",
    "2026-06"
]

base_url = (
    "https://ncua.gov/files/publications/analysis/"
    "call-report-data-{}.zip"
)

data_folder = "data"

os.makedirs(data_folder, exist_ok=True)

for quarter in quarters:
    url = base_url.format(quarter)

    zip_path = os.path.join(
        data_folder,
        f"call-report-data-{quarter}.zip"
    )

    extract_path = os.path.join(
        data_folder,
        f"call-report-data-{quarter}"
    )

    if os.path.exists(extract_path):
        print(f"{quarter}: already downloaded")
        continue

    print(f"Downloading {quarter}...")

    urllib.request.urlretrieve(
        url,
        zip_path
    )

    print(f"Extracting {quarter}...")

    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(extract_path)

    os.remove(zip_path)

print("All quarterly data downloaded.")
```
